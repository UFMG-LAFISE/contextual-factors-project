## A base de dados possui incoerencia do tipo:

- mistura de tipos de dados em uma mesma coluna
- inconsistências de registro



```python
# colunas com dados tipo str()

# mascara booleana novamente
mascara_string = dados_brutos.map(lambda x: isinstance(x, str))

# quais colunas possuem ao menos uma string
colunas_com_string = mascara_string.any()
colunas_alvo = colunas_com_string[colunas_com_string].index.tolist()

print(f"Colunas com strings detectadas: {colunas_alvo}")

# Para cada coluna detectada, mostramos os valores únicos que são strings
for col in colunas_alvo:
    # Filtramos apenas as células que são strings naquela coluna específica
    valores_encontrados = dados_brutos.loc[mascara_string[col], col].unique()
    print(f"-> Na coluna '{col}', encontramos: {valores_encontrados}\n")
```
