```python
def formatar_media_std(x):
    # Calcula a média e desvio, tratando os valores individualmente
    m = x.mean()
    s = x.std()
    # Formata com 1 casa decimal dentro da string (f-string)
    return f"{m:.1f} ± {s:.1f}"

# Aplica o agrupamento
relatorio_final = (
dados_carac[['dits_total', 'zv3', 'dmi', 'dai']].agg(formatar_media_std),
dados_carac.groupby('posicao')[['dits_total', 'zv3', 'dmi', 'dai']].agg(formatar_media_std))

print(relatorio_final)
```
